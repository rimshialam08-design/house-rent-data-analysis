# **Task 2: Data Exploration with Pandas and Python Ecosystem**

```python
# 1. Importing required libraries
import pandas as pd
import numpy as np

# 2. Load dataset
house = pd.read_csv("House_Rent_Dataset.csv")  

# 3. Inspect data
print("--- First 5 Records ---")
display(house.head())

print("--- Dataset Dimensions ---")
print(house.shape)  

print("\n--- Summary of Columns & Data Types ---")
house.info()

print("\n--- Statistical Summary of Numerical Columns ---")
display(house.describe())

print("\n--- Column Names List ---")
print(house.columns)

# 4. Data Cleaning
print("\n--- Checking for Missing/Null Values ---")
print(house.isnull().sum()) 

# Drop missing values if any exist
house.dropna(inplace=True)   

print("\n--- Checking for Duplicate Rows ---")
print(house.duplicated().sum())  

# Drop duplicates if any exist
house.drop_duplicates(inplace=True) 

# 5. Convert Datetime Format

house["Posted On"] = pd.to_datetime(
    house["Posted On"],
    format="%Y-%m-%d"
)

print("\n--- Verification after Datetime Conversion ---")
print(f"Data type of 'Posted On': {house['Posted On'].dtype}")
print(f"Final Cleaned Dataset Shape: {house.shape}")
```

# **Task 3: NumPy Arrays and Array Operations**

In this section, we extract core numerical features from our dataset into native NumPy arrays to inspect array attributes, slicing, reshaping, memory models (copy vs view), and vectorized math.

```python
# Extract target numerical columns into 1D NumPy arrays
rent_arr = house['Rent'].to_numpy()
size_arr = house['Size'].to_numpy()

# 1. Array Creation (Demonstrating 5 distinct creation methods)
arr_from_list = np.array([10, 20, 30, 40, 50])
zeros_mat     = np.zeros((3, 3), dtype=int)
ones_vec      = np.ones((4,), dtype=float)
range_arr     = np.arange(10, 100, 20)
linspace_arr  = np.linspace(0.0, 1.0, 5)

print("--- Array Creation Demonstrations ---")
print(f"arr_from_list: {arr_from_list}")
print(f"zeros_mat (3x3):\n{zeros_mat}")
print(f"ones_vec: {ones_vec}")
print(f"range_arr: {range_arr}")
print(f"linspace_arr: {linspace_arr}")

# 2. Structural Array Properties
matrix_2d = np.column_stack((rent_arr[:10], size_arr[:10]))

print("\n--- Structural Properties of 2D Matrix (Rent, Size) ---")
print(f"shape:    {matrix_2d.shape}")
print(f"ndim:     {matrix_2d.ndim}")
print(f"dtype:    {matrix_2d.dtype}")
print(f"size:     {matrix_2d.size}")
print(f"itemsize: {matrix_2d.itemsize} bytes")

# 3. 1D, 2D, and 3D Indexing and Slicing
print("\n--- Indexing and Slicing ---")
print("1D Slice (Every 500th rent record):", rent_arr[0:2500:500])
print("2D Slice (First 4 rows, Size column only):", matrix_2d[0:4, 1])

# Construct 3D Array: Shape (2 batches, 4 rows, 2 columns)
arr_3d = matrix_2d[:8].reshape(2, 4, 2)
print(f"3D Array Shape: {arr_3d.shape}")
print("3D Slice (Batch 1, all rows, Rent feature):", arr_3d[1, :, 0])

# 4. Reshaping, Transpose, flatten() vs ravel()
sample_block = matrix_2d[:6]  # Shape (6, 2)
reshaped_block = sample_block.reshape(3, 4)
transposed_block = reshaped_block.T

print(f"\nReshaped to (3, 4):\n{reshaped_block}")
print(f"Transposed to (4, 3):\n{transposed_block}")

# Demonstrating deep copy vs memory view
flat_copy = reshaped_block.flatten()  # Deep copy
raveled_view = reshaped_block.ravel()  # Memory view

flat_copy[0] = 999999
raveled_view[1] = 888888

print(f"Original after flatten modification: {reshaped_block[0, 0]} (Unchanged - Deep Copy)")
print(f"Original after ravel modification:   {reshaped_block[0, 1]} (Mutated - Contiguous View)")

# Restore original value
reshaped_block[0, 1] = sample_block[0, 1]

# 5. Vectorized Element-Wise Arithmetic (Zero explicit Python loops)  
price_per_sqft = rent_arr / size_arr
log_rent = np.log1p(rent_arr)

print("\nVectorized Rent per SqFt (first 5 properties):", np.round(price_per_sqft[:5], 2))
```

# **Task 4: Advanced NumPy for EDA**

This section demonstrates broadcasting, row-wise and column-wise aggregations, Boolean filtering, fancy indexing, and reproducible pseudo-random sampling.

```python
# 1. NumPy Broadcasting Demonstration
sample_props = matrix_2d[:5].astype(float)
adjustment_vec = np.array([1000.0, 50.0])  # Shape (2,) broadcast across (5, 2)
adjusted_props = sample_props - adjustment_vec

print("Original Slice (First 2):\n", sample_props[:2])
print("Broadcast Subtracted Slice (First 2):\n", adjusted_props[:2])

# 2. Aggregations & Axis-Based Operations
print("\n--- Summary Aggregations ---")
print(f"Rent Mean:     INR {np.mean(rent_arr):.2f}")
print(f"Rent Std Dev:  INR {np.std(rent_arr):.2f}")
print(f"Rent Variance: {np.var(rent_arr):.2f}")
print(f"Rent Min:      INR {np.min(rent_arr)}")
print(f"Rent Max:      INR {np.max(rent_arr)}")

# Axis 0 (Column-wise across all records) vs Axis 1 (Row-wise across features)
col_means = np.mean(matrix_2d, axis=0)
row_maxima = np.max(matrix_2d[:5], axis=1)
print(f"Column Means (Axis 0) [Mean Rent, Mean Size]: {col_means}")
print(f"Row Maxima for first 5 properties (Axis 1): {row_maxima}")

# 3. Boolean and Fancy Indexing
luxury_mask = rent_arr > 200000
luxury_listings = rent_arr[luxury_mask]
print(f"\nCount of properties with Rent > 200k: {len(luxury_listings)}")

# Fancy Indexing: Non-contiguous extraction using arbitrary index positions
full_matrix_2d = np.column_stack((rent_arr, size_arr))
target_indices = np.array([5, 42, 108, 512, 1024])
fancy_sample = full_matrix_2d[target_indices]
print(f"\nFancy Indexed Records from positions {target_indices}:\n{fancy_sample}")

# 4. Reproducible Random Sampling
rng = np.random.default_rng(seed=42)
sampled_idx = rng.choice(len(rent_arr), size=5, replace=False)
reproducible_sample = rent_arr[sampled_idx]

print(f"\nDeterministic Random Sample (Seed 42): {reproducible_sample}")
```

# **Task 6: EDA Communication, Findings & Conclusions**

### Summary of Key Findings
1. **Geographic Pricing Dominance:** Metro tiering is the strongest predictor of rental costs; physical size alone does not govern valuation without geographic context.
2. **Furnishing Value Premium:** Furnished properties maintain a distinct price premium over unfurnished units across all configurations.
3. **Severe Distribution Asymmetry:** Monthly rents follow a heavy right-tailed distribution, making median and interquartile metrics more representative of baseline market behavior than arithmetic means.

---

### Answering the Central Analytical Question
The exploratory analysis confirms that while property dimensions (`Size`) and configurations (`BHK`) establish baseline accommodations, metropolitan location (`City`) dictates overall price-per-square-foot elasticity, and `Furnishing Status` provides a consistent upward baseline shift.

---

### Distinguishing Observation from Causation
While furnished units show higher rents, this correlation does not prove that adding furniture causes rent to rise by that exact amount. Furnished accommodations are more common in affluent, prime-location buildings, introducing location-based confounding factors.

---

### Dataset Limitations
1. **Lack of Neighborhood-Level Coordinates:** Listings are categorized only by broad city names rather than micro-markets, zip codes, or proximity to transit hubs, which masks local neighborhood variations.
2. **Absence of Longitudinal Tracking:** The dataset is a cross-sectional snapshot rather than a multi-year series, preventing the evaluation of inflation and seasonal rental trends.

---

### Recommended Follow-Up Analyses
* **Spatial Hedonic Regression:** Fit regression models incorporating distance-to-center metrics and transit proximity to isolate neighborhood premiums.
* **Unsupervised Tenant Clustering:** Use K-Means clustering across normalized size, bathroom ratio, and rent values to group properties into distinct market personas.

---

### Conclusion
This exploratory data analysis establishes that metropolitan city tier and furnishing status are the primary drivers of rental valuations, rather than physical square footage alone. The NumPy implementations also demonstrate the utility of vectorized operations, multidimensional indexing, broadcasting, and reproducible random sampling for data science workflows.
